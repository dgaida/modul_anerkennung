## 🚀 Beispiel: Modulvergleich
Hier initialisieren wir das Tool und führen einen Vergleich durch.

```python
from modul_anerkennung.llm_interface import LLMInterface
from modul_anerkennung.rag_manager import RAGManager
from modul_anerkennung.similarity_checker import SimilarityChecker

# Komponenten initialisieren
rag = RAGManager()
llm = LLMInterface()
checker = SimilarityChecker(rag, llm)

external_module_desc = (
    "Grundlagen der Informatik: Einführung in Programmierung und Datenstrukturen."
)


async def run_comparison():
    # Optional: Ein eigenes Modulhandbuch indexieren
    # await rag.process_document("mein_handbuch.pdf")

    result = await checker.compare_modules(external_module_desc)
    return result


# In Colab/Jupyter nutzen wir direkt await
result = await run_comparison()
print(result)
```
